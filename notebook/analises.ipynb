```python
import pandas as pd

# Carregar o arquivo JSON em pedaços menores (chunks)
chunk_size = 100000  # Ajuste conforme necessário
chunks = pd.read_json('order.json', lines=True, chunksize=chunk_size)

# Criar uma lista para armazenar os pedaços de dados
df_chunks = []
for chunk in chunks:
    # Selecionar apenas as colunas necessárias para economizar memória
    chunk = chunk[['customer_id', 'order_id', 'order_total_amount', 'order_created_at']]

    # Adicionar o chunk ao DataFrame final
    df_chunks.append(chunk)

# Concatenar todos os pedaços de dados em um único DataFrame
df_order = pd.concat(df_chunks, ignore_index=True)

# Verificar as primeiras linhas para garantir que tudo foi carregado corretamente
print(df_order.head())
```

```python
# Juntar os dados de pedidos com as informações de A/B Test
df_order_ab = pd.merge(df_order, df_ab_test[['customer_id', 'is_target']], on='customer_id', how='left')

# Verificar as primeiras linhas para garantir que a junção foi feita corretamente
print(df_order_ab.head())
```

```python
# Calcular o ticket médio por grupo (controle vs target)
ticket_por_grupo = df_order_ab.groupby('is_target')['order_total_amount'].mean()

# Exibir o resultado
print("Ticket Médio por Grupo (Controle vs Target):")
print(ticket_por_grupo)
```

```python
# Número total de pedidos por usuário por grupo
retencao_por_usuario = df_order_ab.groupby(['customer_id', 'is_target'])['order_id'].count().reset_index()

# Usuários com mais de 1 pedido (retenção)
retencao = retencao_por_usuario[retencao_por_usuario['order_id'] > 1]

# Taxa de retenção por grupo
retencao_por_grupo = retencao.groupby('is_target')['customer_id'].nunique() / df_order_ab.groupby('is_target')['customer_id'].nunique()
print("Taxa de Retenção por Grupo (Controle vs Target):")
print(retencao_por_grupo)
```

```python
from scipy.stats import ttest_ind

# Filtrar os dados por grupo
group_target = df_order_ab[df_order_ab['is_target'] == 'target']['order_total_amount']
group_control = df_order_ab[df_order_ab['is_target'] == 'control']['order_total_amount']

# Aplicar o teste t para comparar o ticket médio entre os grupos
t_stat, p_value = ttest_ind(group_target, group_control)

print(f"Estatística t (Ticket Médio): {t_stat}")
print(f"P-valor (Ticket Médio): {p_value}")

# Verificar se a diferença é estatisticamente significativa
if p_value < 0.05:
    print("✅ A diferença é estatisticamente significativa no ticket médio.")
else:
    print("⚠️ A diferença no ticket médio não é estatisticamente significativa.")
```

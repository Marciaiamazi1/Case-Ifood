```python
# Simulação de viabilidade financeira da campanha de cupons

# Definições iniciais (valores hipotéticos que podem ser ajustados conforme necessário)
ticket_medio_target = 47.8  # Ticket médio do grupo que recebeu o cupom (Target)
ticket_medio_controle = 42.5  # Ticket médio do grupo controle (sem cupom)
pedidos_target = 12000  # Número total de pedidos feitos no grupo Target
custo_por_cupom = 10  # Custo estimado por pedido com cupom

# 1. Estimação do ganho por pedido
ganho_por_pedido = ticket_medio_target - ticket_medio_controle

# 2. Cálculo do ganho total
ganho_total = ganho_por_pedido * pedidos_target

# 3. Cálculo do custo total da campanha
custo_total = pedidos_target * custo_por_cupom

# Exibição dos resultados
print("📊 Avaliação de Viabilidade Financeira")
print("---------------------------------------")
print(f"Ticket médio do grupo Target: R$ {ticket_medio_target}")
print(f"Ticket médio do grupo Controle: R$ {ticket_medio_controle}")
print(f"Ganho médio por pedido: R$ {ganho_por_pedido:.2f}")
print(f"Número de pedidos no grupo Target: {pedidos_target}")
print(f"Ganho total estimado: R$ {ganho_total:,.2f}")
print(f"Custo total da campanha: R$ {custo_total:,.2f}")

# 4. Verificação da viabilidade financeira
if ganho_total > custo_total:
    print("\n✅ Campanha financeiramente viável.")
else:
    print("\n⚠️ Campanha não se pagou com os resultados estimados.")
```

```python
import matplotlib.pyplot as plt

# Dados fornecidos
custo_por_cupom = 10  # R$ 10
pedidos_target = 12000  # 12.000 pedidos
ganho_medio_por_pedido = 5.30  # R$ 5,30
receita_adicional = 63600  # R$ 63.600
custo_total_campanha = 120000  # R$ 120.000

# Calcular o custo total
custo_total = pedidos_target * custo_por_cupom

# Calcular o ROI
ROI = receita_adicional / custo_total_campanha

# Calcular o custo por pedido
custo_por_pedido = custo_total_campanha / pedidos_target

# Criar o gráfico de comparação entre custo e receita adicional
plt.figure(figsize=(10, 6))
labels = ['Custo Total', 'Receita Adicional']
values = [custo_total, receita_adicional]
plt.bar(labels, values, color=['red', 'blue'])
plt.title('Comparação entre Custo Total e Receita Adicional')
plt.ylabel('Valor (R$)')
plt.show()

# Criar o gráfico de comparação entre custo por pedido e ganho médio por pedido
plt.figure(figsize=(10, 6))
labels = ['Custo por Pedido', 'Ganho Médio por Pedido']
values = [custo_por_pedido, ganho_medio_por_pedido]
plt.bar(labels, values, color=['orange', 'green'])
plt.title('Comparação entre Custo por Pedido e Ganho Médio por Pedido')
plt.ylabel('Valor (R$)')
plt.show()

# Gráfico de ROI
plt.figure(figsize=(8, 5))
plt.bar(['ROI'], [ROI], color='purple')
plt.title('ROI da Campanha')
plt.ylabel('ROI')
plt.ylim(0, 2)  # Limitar o gráfico de ROI entre 0 e 2
plt.show()
```

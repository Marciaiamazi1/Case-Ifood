```python
import urllib.request
import tarfile
import pandas as pd
import json
import os
import gzip # Import the gzip library

# URLs fornecidas para baixar os arquivos
urls = {
    "order.json.gz": "https://data-architect-test-source.s3-sa-east-1.amazonaws.com/order.json.gz",
    "consumer.csv.gz": "https://data-architect-test-source.s3-sa-east-1.amazonaws.com/consumer.csv.gz",
    "restaurant.csv.gz": "https://data-architect-test-source.s3-sa-east-1.amazonaws.com/restaurant.csv.gz",
    "ab_test.tar.gz": "https://data-architect-test-source.s3-sa-east-1.amazonaws.com/ab_test_ref.tar.gz"  # URL corrigida
}

# Baixar os arquivos
for nome_arquivo, url in urls.items():
    print(f"⬇️ Baixando {nome_arquivo}...")
    urllib.request.urlretrieve(url, f"/content/{nome_arquivo}")

print("\n✅ Todos os arquivos foram baixados para /content/")

# Descompactar o arquivo .tar.gz (ab_test)
file_path = '/content/ab_test.tar.gz'
with tarfile.open(file_path, "r:gz") as tar:
    tar.extractall(path="/content/")

# Verificar os arquivos descompactados
print("\nArquivos descompactados:")
print(os.listdir('/content/'))

# Carregar os arquivos CSV (para consumers e restaurants)
df_consumer = pd.read_csv('/content/consumer.csv.gz')
df_restaurant = pd.read_csv('/content/restaurant.csv.gz')

# Carregar arquivo JSON (para orders)
with gzip.open('/content/order.json.gz', 'rt', encoding='utf-8') as f: # Use gzip.open to read the gzipped file
    df_order = pd.json_normalize(json.load(f))

# Carregar arquivo CSV de A/B Test (os arquivos descompactados)
df_ab_test = pd.read_csv('/content/ab_test_ref.csv')  # Supondo que o arquivo CSV tenha sido descompactado corretamente

# Exibir as primeiras linhas de cada DataFrame para verificação
print("Dados do Consumer:")
print(df_consumer.head())
print("Dados do Restaurant:")
print(df_restaurant.head())
print("Dados do Order:")
display(df_order.head()) # Use display() for better formatting
print("Dados do A/B Test:")
print(df_ab_test.head())
```
